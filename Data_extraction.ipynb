## Lambda Function

```python
# basic function

def sum_values(a,b):
    x = a+b
    return x
```

```python
sum_values(2,3)
```

```python
#lambda function

sum_values_lambda = lambda a,b : a+b
sum_values_lambda(2,3)
```

```python
sum_values
```

```python
sum_values_lambda
```
