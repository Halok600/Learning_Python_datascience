```python
import re
```

```python
text = """Hello World.
I love Python!!!!!
My first lucky number is 987-654-321
My second lucky number is 876-543-210
My third lucky number is 765-432-100
My fourth lucky number is 123-456-123-123-456
"""
```

## re.search()


```python
re.search(r'\d',text)
```

```python
re.search(r"^Hello",text,flags =re.M)
```

### re.findall()

```python
re.findall(r'\d',text)
```

```python
re.findall(r'[^\w\s]',text)
```

```python
text1 = ''' 
            13-04-2021
            2021-13-04
            2021-04-13
            04-13-2021'''
```

```python
re.findall(r'\d{2}-\d{2}-\d{4}',text1)
```

```python
text2 = '''
    username10
    abc
    10
    halok600
    '''
```

```python
re.findall(r'[a-zA-Z0-9]{4,14}',text2)
```
