### The Split-Apply-Combine Strategy

```python
#split the data into separate groups
passenger_filter = df_cars['Vehicle_type']=='Passenger'
car_filter = df_cars['Vehicle_type']=='Car'
#df_cars['Vehicle_type'].value_counts()
```

```python
car_filter
```

```python
passenger_filter
```

```python
#apply an operation
passeneger_avg = df_cars[passenger_filter]['Sales_in_thousands'].mean()
car_avg = df_cars[car_filter]['Sales_in_thousands'].mean()
```

```python
pd.DataFrame({'Vehicle_type':['Car','Passenger'],
              'Sales_in_thousands':[car_avg,passeneger_avg]}).set_index('Vehicle_type')
```

### The filter() method

```python
df_cars.mean(numeric_only=True)
```

```python
df_filter = df_cars.groupby(['Manufacturer']).filter(filter_fun)
df_filter
```
